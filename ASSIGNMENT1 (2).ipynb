```python
# Mean, Median and Mode for Grouped Data

# Age groups
age_groups = [(15, 25), (25, 35), (35, 45), (45, 55), (55, 65)]

# Number of persons
frequency = [350, 375, 120, 80, 90]

# --------------------------------------------------
# 1. MEAN


# Calculate class midpoints
midpoints = [(lower + upper) / 2 for lower, upper in age_groups]

# Calculate f*x
fx = [f * x for f, x in zip(frequency, midpoints)]

# Mean = Σfx / Σf
total_frequency = sum(frequency)
mean = sum(fx) / total_frequency

print("Total number of persons =", total_frequency)
print("Mean =", round(mean, 2))




```
